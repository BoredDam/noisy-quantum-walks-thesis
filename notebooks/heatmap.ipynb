```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import PowerNorm

steps = 100
max_pos = 100
noise_odds = 0.5
seed = 42

fig, axes = plt.subplots(
    1,
    2,
    figsize=(14, 5),
    sharex=True,
    sharey=True,
    constrained_layout=True
)

for ax, noise_type in zip(axes, ["Z", "X"]):

    Y, X_positions = noisy_circle_quantum_random_walk_1D(
        max_pos=max_pos,
        steps=steps,
        odds=0.5,
        noise_type=noise_type,
        symmetrical=True,
        noise_odds=noise_odds,
        seed=seed
    )

    image = ax.imshow(
        Y,
        cmap="inferno",
        origin="lower",
        norm=PowerNorm(gamma=0.38),
        extent=[
            X_positions[0],
            X_positions[-1],
            0,
            steps
        ],
        aspect="auto"
    )

    ax.set_title(
        f"Canale {noise_type}, p={noise_odds}"
    )
    ax.set_xlabel("posizione")

axes[0].set_ylabel("tempo")

fig.suptitle(
    "Passeggiata aleatoria quantistica con rumore, confronto tra Z-dephasing e X-dephasing",
    fontsize=14
)


plt.savefig(
     f"./images/confronto_X_Z_noise_{str(noise_odds).replace('.', '_')}_seed_{seed}.pdf",
     bbox_inches="tight",
     pad_inches=0
)

plt.show()
```
